```python
from langchain_core.tools import tool
import requests
from langchain_core.messages import HumanMessage
from langchain_core.tools import InjectedToolArg
from typing import Annotated


@tool
def get_conversion(base_currency: str, target_currency: str) -> dict:
  """this function fetches the currency conversion between a given base currency and a target currency
  """
  url = f'https://v6.exchangerate-api.com/v6/84754a833bec424acf4c5f5e/pair/{base_currency}/{target_currency}'
  response = requests.get(url)
  data = response.json()
  return data
@tool
def convert(base_currency_value:int, conversion_rate:Annotated[float,InjectedToolArg])->float:
  """
  given a currency conversion rate this function calculates the target currency value from a given base currency value
  """
  return base_currency_value*conversion_rate
```

```python
get_conversion.invoke({'base_currency':'USD','target_currency':'INR'})
```

```python
convert.invoke({'base_currency_value': 456, 'conversion_rate': 456.98})
```

# **Tool Binding**

```python
llm=ChatGoogleGenerativeAI(model='gemini-3.5-flash', temperature=0.2)
```

```python
llm_tool=llm.bind_tools([get_conversion,convert])
```

```python
messages=[HumanMessage("what is the conversin factor between USD and INR , and based on that convert 456 USD to INR")]
```

```python
messages
```

```python
ai_message=llm_tool.invoke(messages)
```

```python
ai_message.tool_calls
```

```python
ai_message
```

```python
import json

for tool_call in ai_message.tool_calls:
  # execute the 1st tool and get the value of conversion rate
  if tool_call['name'] == 'get_conversion_factor':
    tool_message1 = get_conversion.invoke(tool_call)
    # fetch this conversion rate
    conversion_rate = json.loads(tool_message1.content)['conversion_rate']
    # append this tool message to messages list
    messages.append(tool_message1)
  # execute the 2nd tool using the conversion rate from tool 1
  if tool_call['name'] == 'convert':
    # fetch the current arg
    tool_call['args']['conversion_rate'] = conversion_rate
    tool_message2 = convert.invoke(tool_call)
    messages.append(tool_message2)
```
